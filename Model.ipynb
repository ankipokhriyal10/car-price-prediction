Data Analysis

```python
for col in cars_data.columns:
    print('Unique values of ' + col)
    print(cars_data[col].unique())
    print("======================")
```

```python
def get_brand_name(car_name):
    car_name = car_name.split(' ')[0]
    return car_name.strip()
```

```python
def clean_data(value):
    value = value.split(' ')[0]
    value = value.strip()
    if value == '':
        value = 0
    return float(value)
```

```python
get_brand_name('Maruti  Swift Dzire VDI')
```

```python
cars_data['name'] = cars_data['name'].apply(get_brand_name)
```

```python
cars_data['name'].unique()
```

```python
cars_data['mileage'] = cars_data['mileage'].apply(clean_data)
```

```python
cars_data['max_power'] = cars_data['max_power'].apply(clean_data)
```

```python
cars_data['engine'] = cars_data['engine'].apply(clean_data)
```

```python
for col in cars_data.columns:
    print('Unique values of ' + col)
    print(cars_data[col].unique())
    print("======================")
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
cars_data['name'] = cars_data['name'].replace({
    'Maruti': 1, 'Skoda': 2, 'Honda': 3, 'Hyundai': 4, 'Toyota': 5,
    'Ford': 6, 'Renault': 7, 'Mahindra': 8, 'Tata': 9,
    'Chevrolet': 10, 'Datsun': 11, 'Jeep': 12, 'Mercedes-Benz': 13,
    'Mitsubishi': 14, 'Audi': 15, 'Volkswagen': 16, 'BMW': 17,
    'Nissan': 18, 'Lexus': 19, 'Jaguar': 20, 'Land': 21, 'MG': 22,
    'Volvo': 23, 'Daewoo': 24, 'Kia': 25, 'Fiat': 26, 'Force': 27,
    'Ambassador': 28, 'Ashok': 29, 'Isuzu': 30, 'Opel': 31
})
```

```python
cars_data['transmission'].unique()
```

```python
cars_data['transmission'] = cars_data['transmission'].replace({
    'Manual': 1,
    'Automatic': 2
})
```

```python
cars_data['seller_type'].unique()
```

```python
cars_data['seller_type'] = cars_data['seller_type'].replace({
    'Individual': 1,
    'Dealer': 2,
    'Trustmark Dealer': 3
})
```

```python
cars_data.info()
```

```python
cars_data['fuel'].unique()
```

```python
cars_data['fuel'] = cars_data['fuel'].replace({
    'Diesel': 1,
    'Petrol': 2,
    'LPG': 3,
    'CNG': 4
}) 
```

```python
cars_data.info()
```

```python
cars_data.reset_index(inplace=True)
```

```python
cars_data
```

```python
cars_data['owner'].unique()
```

```python
cars_data['owner'] = cars_data['owner'].replace({
    'First Owner': 1,
    'Second Owner': 2,
    'Third Owner': 3,
    'Fourth & Above Owner': 4,
    'Test Drive Car': 5
})
```

```python
cars_data.drop(columns=['index'], inplace=True)
```

```python
input_data = cars_data.drop(columns=['selling_price'])
output_data =cars_data['selling_price']
```

```python
train_x,test_x,train_y,test_y = train_test_split(input_data, output_data, test_size=0.2)
```

Model Creation

```python
model = LinearRegression()
```

Train Model
